# Titanic Survival Prediction with Logistic Regression

This project predicts whether a passenger survived the Titanic disaster using machine learning.

## 1. Import Libraries

We import NumPy and Pandas for data handling, Seaborn for the Titanic dataset, Matplotlib for visualization, and warning controls.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

## 2. Load the Dataset

The Titanic dataset is loaded from Seaborn. Each row represents one passenger.

```python
df = sns.load_dataset('titanic')
df.head()
```

## 3. Explore the Dataset

We inspect the column names, data types, and missing values before preparing the data.

```python
df.columns
```

```python
df.info()
```

```python
df.drop(['deck', 'embark_town','alive','class', 'who', 'adult_male'], axis = 1, inplace = True)
```

```python
df.info()
```

## 4. Data Preprocessing

Unnecessary columns are removed. Missing ages are replaced with the average age, and rows with missing embarkation values are removed.

```python
df["age"].fillna(df["age"].mean(), inplace = True)
```

it select the age colomn and replace all its null values with the mean of age column

```python
df.dropna(subset=["embarked"], inplace=True)
```

```python
df.info()
```

## 5. Convert Text into Numbers

Machine-learning models require numerical input. Therefore, the `sex` and `embarked` columns are converted into numerical labels.

```python
import sklearn.preprocessing
le = sklearn.preprocessing.LabelEncoder()
df['sex'] = le.fit_transform(df['sex'])
df['embarked'] = le.fit_transform(df['embarked']) # S=2, C=0, Q=1
```

```python
df = df.astype(int)
```

```python
df
```

## 6. Separate Features and Target

`X` contains the passenger information used for prediction.  
`y` contains the target value:

- `0` = Did not survive
- `1` = Survived

```python
x = df.drop('survived', axis = 1)
y = df['survived']
```

## 7. Split the Dataset

The data is divided into training and testing sets. The model learns from the training data and is evaluated on unseen test data.

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

## 8. Train the Logistic Regression Model

Logistic Regression is a classification algorithm used to predict one of two outcomes: survived or did not survive.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
model = LogisticRegression()
model.fit(x_train, y_train)
```

## 9. Make Predictions

The trained model predicts survival outcomes for passengers in the test dataset.

```python
y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
y_test
```

## 10. Evaluate the Model

Accuracy shows the percentage of correct predictions. The confusion matrix and classification report provide additional performance details.

```python
accuracy_score(y_test, y_pred)
```

```python
confusion_matrix(y_test, y_pred)
```

```python
print(classification_report(y_test, y_pred))
```
