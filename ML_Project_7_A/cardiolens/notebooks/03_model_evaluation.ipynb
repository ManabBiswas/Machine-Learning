# 1. Import libraries

```python
import pandas as pd
import numpy as np
import joblib
from sklearn.metrics import accuracy_score, f1_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

```

# 2. Load test data

```python
x_test_processed = pd.read_csv("../backend/data/processed/x_test_processed.csv").values
y_test = pd.read_csv("../backend/data/processed/y_test.csv").squeeze().values
```

# 3. Load saved model

```python
model = joblib.load("../backend/models/logistic_regression_model.pkl")
```

# 4. Predict

```python
y_pred = model.predict(x_test_processed)
```

# 5. Compute metrics


```python
print("Accuracy:", accuracy_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

# 6. Plot confusion matrix

```python
cm = confusion_matrix(y_test, y_pred)

sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=["No Disease", "Disease"],
            yticklabels=["No Disease", "Disease"])

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```

### conclusion

```python
tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel()
print(f"The Logistic Regression model achieved an accuracy of {accuracy_score(y_test, y_pred):.1%} and an F1 score of {f1_score(y_test, y_pred):.1%}.")
print(f"The confusion matrix shows {tn} true negatives, {tp} true positives, and {fp + fn} incorrect predictions.")
print("Labels: 1 = disease present, 0 = no disease.")
```
